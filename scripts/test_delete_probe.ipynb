```python
lverbose = 0
for indx, fl_name in enumerate(ntk.natural_sort(glob.glob('/Scratch/kiranbn/test_delete_probe/ZBR00054_L7_W3_2023-05-09_12-55-18/Headstages_192_*.bin'))):
        print(f'\n{fl_name}') 

        ## check probe 0
        
        # fl_name = '/Scratch/kiranbn/test_delete_probe/ZBR00054_L7_W3_2023-05-09_12-55-18/Headstages_192_Channels_int16_2023-05-09_21-40-20.bin'
        number_of_channels = 192
        hstype = ['linear','linear', 'linear']
        nprobes = 3
        probenum = 0
        probechans = 64
        ts = 0
        te = 25000
        t, dgc = ntk.load_raw_gain_chmap_1probe(fl_name, number_of_channels,
                                                hstype, nprobes=nprobes,
                                                lraw=1, ts=ts, te=te,
                                                probenum=probenum,
                                                probechans=probechans)
        print(f'{fl_name.replace("Headstages_192_Channels_int16_", "Headstages_128_Channels_int16_")}') 
        # fl_name_new = '/Scratch/kiranbn/test_delete_probe/ZBR00054_L7_W3_2023-05-09_12-55-18/Headstages_128_Channels_int16_2023-05-09_21-40-20.bin'
        fl_name_new = fl_name.replace("Headstages_192_Channels_int16_", "Headstages_128_Channels_int16_")
        number_of_channels = 128
        hstype = ['linear','linear']
        nprobes = 2
        probenum = 0
        probechans = 64
        ts = 0
        te = 25000
        t_new, dgc_new = ntk.load_raw_gain_chmap_1probe(fl_name_new, number_of_channels,
                                                hstype, nprobes=nprobes,
                                                lraw=1, ts=ts, te=te,
                                                probenum=probenum,
                                                probechans=probechans)
        if lverbose:
            print(f't {t} t_new {t_new}')
            print(f'dgc {dgc}')
            print(f'dgc_new {dgc_new}')
        
        if not np.isclose(t, t_new):
            raise ValueError(f'Error: {fl_name} {t} {t_new}')
        if lverbose:
            print(f'shape {np.isclose(dgc, dgc_new).flatten().shape}')
            print(f'len True {np.sum(np.isclose(dgc, dgc_new).flatten())}')
        if not np.allclose(dgc, dgc_new):
            raise ValueError(f'Error: {fl_name} {dgc} {dgc_new}')


    
```
